## Pearsons Correlation Coefficient
* Measure of the linear relationship between two features 
    * Ratio of the covariance of x & y to the product of their standard deviation
    * takes pairs = each pair is an observation
    * -1 to 1
    * =1 --> perfect linear relationship
    * \>0 --> positive correlation
    * =0 --> no linear relationship
    * <0 --> negative correlation
    * =-1 --> perfect -ve linear relationship.


```python
#Show the correlation coefficient between each trait between two judges
def corrPerTrait(judgeA, judgeB):
    listOfCoefficients = []
    i=1
    while i <6:
        listOfCoefficients.append(np.corrcoef(judgeA[:,i].tolist(), judgeB[:,i].tolist())[0:1][0][1])
        i = i + 1
    return (listOfCoefficients)

# corrPerTrait(judge1,judge2)
```

## Spearman Rank & Spearman Brown 

Spearman Rank Correlation Between judges

* Spearman Rank Correlation - a nonparametric version of the Pearon correlation. It measures the strength and direction of association between two ranked variables.
    * Spearman Rank Correlation is used to measure the correlation between two ranked variables.
    * The funcion first ranks the data by sorting it in ascending order, then it assigns a rank to based on the position of the data and then replaces the original points with their respective ranks
    * Then the pearson correlation coefficient of the ranks is found

* Output - Spearman rank correlation (rho) and corresponding p-value
    * Rho -> -1 to 1
    * Two-sided p value -> the probability that the observed rho correlation occured by chance. Assuming that the null hypothesis is true - the two variables are uncorrelated. - A low p-value suggests that the pbserved correlation is statistically significant.
        * p-value < 0.05 - observed correlation is unlikely to be a result of chance
        * large p-value - correlation could plausibly be random chance

```python
#Cornbach's Aplha 
# measures internal consistency/reliability of a set of survey items
# high values = higher level of agreement between items (0-1 scale)
# high values = responses 

#output - alpha & confidence interval
def cronbach_alpha(trait):
    data = {}
    for i in range(11):
        key = f'Judge {i+1}'
        if (i<9):
            x = '0' + str(i+1)
        else:
            x = str(i+1)
        value = np.array(printJudge(x))[:,trait].tolist()
        data[key] = value
    alpha = pg.cronbach_alpha(data=pd.DataFrame(data))
    return alpha
    # return (n * alpha) / (1 + (n - 1) * alpha)

print(cronbach_alpha(3))

def cronbach_alpha_traits(judge):
    data = {}
    for i in range(5):
        key = f'trait {i+1}'
        value = np.array(printJudge(judge))[:,i+1].tolist()
        data[key] = value
    alpha = pg.cronbach_alpha(data=pd.DataFrame(data))
    return alpha
    # return (n * alpha) / (1 + (n - 1) * alpha)

print(cronbach_alpha_traits('02'))

def spearmanBrown(trait, numberOfjudges):
    old = cronbach_alpha(trait)[0]
    return (numberOfjudges * old / (1 + (numberOfjudges - 1) * old))

def spearmanBrownTrait(judge, numberOfTraits):
    old = cronbach_alpha_traits(judge)[0]
    return (numberOfTraits * old / (1 + (numberOfTraits - 1) * old))

print(spearmanBrownTrait('02', 7))

########WHY IS THE RETURN OVER 1???????
```
